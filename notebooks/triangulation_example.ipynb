```python
import numpy as np
from matplotlib.pyplot import show

from pwlopt.fitting.triangulation import Triangulation
```

```python
pts = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])
```

```python
triangulation = Triangulation.from_points(pts)
```

```python
triangulation.trg[0].sample(.05)
```

```python
triangulation.trg[0].plot_sample()
show()
```

```python
new_pt = np.array([0.25, .35])
triangulation.insert_point(new_pt)
```

```python
fig, ax = triangulation.plot_triangulation()
triangulation.trg[0].sample(.1)
triangulation.trg[3].sample(.05)
triangulation.trg[0].plot_sample(fig, ax)
triangulation.trg[3].plot_sample(fig, ax)
show()
```

```python
new_pt = np.array([0.75, .35])
triangulation.insert_point(new_pt)
```

```python
fig, ax = triangulation.plot_triangulation()
triangulation.trg[0].sample(.1)
triangulation.trg[3].sample(.05)
triangulation.trg[1].sample(.075)
triangulation.trg[0].plot_sample(fig, ax)
triangulation.trg[3].plot_sample(fig, ax)
triangulation.trg[1].plot_sample(fig, ax)
show()
```
