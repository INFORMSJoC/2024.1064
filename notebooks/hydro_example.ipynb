```python
import logging

from pwlopt.hydropower.hydro_data import HydroData
```

```python
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] | %(name)s | %(message)s",
    force=True,
)
```

```python
data_path = "../data/hydro_instances"
month = "April"
plantfile = f"{data_path}/{month}/plant.json"
inflowfile = f"{data_path}/{month}/inflow.json"
pricefile = f"{data_path}/{month}/price.json"
pumpcostfile = f"{data_path}/{month}/pump_cost.json"
```

```python
trg_path = "../data/triangulations/non-grid"
size = "small"
inst = 0
trg_file = f"{trg_path}/{size}/inst{inst}.txt"
```

```python
data = HydroData.from_file(
    plantfile=plantfile,
    inflowfile=inflowfile,
    pricefile=pricefile,
    pumpcostfile=pumpcostfile
)
```

```python
len(data.price)
```

```python
from pwlopt.hydropower.hpf_pwl.hydro_cc import HydroCC

model = HydroCC(data)
model.readTrg(trg_file)
model.buildModel()
model.solve()
```

```python
from pwlopt.hydropower.hpf_pwl.hydro_dcc import HydroDCC

model = HydroDCC(data)
model.readTrg(trg_file)
model.buildModel()
# model.solve()
```

```python
from pwlopt.hydropower.hpf_pwl.hydro_mc import HydroMC

model = HydroMC(data)
model.readTrg(trg_file)
model.buildModel()
# model.solve()
```

```python
from pwlopt.hydropower.hpf_pwl.hydro_dlog import HydroDLog

model = HydroDLog(data)
model.readTrg(trg_file)
model.buildModel()
# model.solve()
```

```python
from pwlopt.hydropower.hpf_pwl.hydro_inc import HydroInc

model = HydroInc(data)
model.readTrg(trg_file)
model.buildModel()
# model.solve()
```

```python
from pwlopt.hydropower.hpf_pwl.hydro_gib import HydroGIB

model = HydroGIB(data)
model.readTrg(trg_file)
model.buildPWL()
model.buildModel()
# model.solve()
```

```python
from numpy.random import randint

from pwlopt.hydropower.hpf_pwl.hydro_log_grid import GridTriangulation, HydroLogGrid

grid_trg = GridTriangulation(randint(2, size=(4, 4)), 4, 4, 0, 1, 0, 1)

model = HydroLogGrid(data)
model.setGridTrg(grid_trg)
model.buildModel()
model.solve()
```
